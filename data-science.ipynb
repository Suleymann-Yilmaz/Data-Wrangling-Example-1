## E- Aykırı Değer
*# Aykırı Değer veri setinde normal dağılımdan ayrı yönde bulunan aykırı değerdir. Aykırı değer hata değildir. Değerin sıradışı olmasıyla ortaya çıkan durumlar incelenmelidir*#

```python
df[sayısal_sutunlar]
```

```python
df.dtypes
```

```python
df[sayısal_sutunlar].describe()
```

```python
Q1_birim = np.quantile(df["birim_fiyat"],0.25)
Q3_birim = np.quantile(df["birim_fiyat"],0.75)
```

```python
Q1_birim,Q3_birim
IQR_birim = Q3_birim - Q1_birim
upper_birim = Q3_birim + 1 * IQR_birim
lower_birim = Q1_birim - 1 * IQR_birim

count_of_outlier_unit=df[(df["birim_fiyat"]>=lower_birim)&(df["birim_fiyat"]<=upper_birim)]
len(df)-len(count_of_outlier_unit)
```

* Birim fiyatta 162 tane outlier yani aykırı değer bulunmaktadır.

```python
aykırı_değerler = {}
for sutun in sayısal_sutunlar:
    Q1 = df[sutun].quantile(0.25)
    Q3 = df[sutun].quantile(0.75)
    IQR = Q3-Q1
    alt_sınır = Q1-1.5*IQR
    ust_sınır = Q3+1.5*IQR

    aykırı_değerler[sutun] = {"Q1":Q1,"Q3":Q3,"IQR":IQR,"alt_sınır":alt_sınır,"ust_sınır":ust_sınır}

aykırı_değerler
```

```python
aykırı_maskesi = pd.Series(False,index = df.index)

for sutun in sayısal_sutunlar:
    alt_sınır = aykırı_değerler[sutun]["alt_sınır"]
    ust_sınır = aykırı_değerler[sutun]["ust_sınır"]

    sutun_maskesi = (df[sutun]<alt_sınır) | (df[sutun]>ust_sınır)
    aykırı_maskesi = aykırı_maskesi | sutun_maskesi
```

```python
aykırı_maskesi = pd.DataFrame(aykırı_maskesi)
aykırı_maskesi.value_counts()
```

# N - Veri Hikayeleştirme ve İş Yorumu

Amaç:
  1. Gerçek hayata benzeyen iş sorularını veri ile cevaplamak
  2. Analiz sonuçlarını tablo, grafik ve kısa iş yorumları ile sunmak

Adımlar:
  1. Hangi kategori hem yüksek gelir üretiyor hem de müşteri memnuniyeti açısından güçlü görünüyor?
  2. Hangi ödeme türleri daha yüksek ortalama sipariş tutarı üretiyor ve müşteri puanı açısından performans gösteriyor?
  3. Hangi şehir-kategori kombinasyonları en güçlü fırsatı sunuyor?

```python
df.head()
```

```python
# hangi kategori hem yüksek puan hem de yüksek müşteri memmuniyeti sağlıyor?
kategori_raporu = df.groupby("kategori").agg(
    toplam_gelir =("toplam_tutar","sum"),
    ortalama_musteri_puani = ("musteri_puani","mean")
).round(2)
kategori_raporu
```

```python
kategori_raporu = kategori_raporu.sort_values(by = "toplam_gelir",ascending = False)
kategori_raporu
```

```python
# Kategoriye göre toplam gelir
plt.Figure(figsize  = (15,10))
sns.barplot(data = kategori_raporu,x = "kategori",y = "toplam_gelir")
plt.title("Kategoriye göre toplam gelir")
plt.xlabel("Kategori")
plt.ylabel("Toplam gelir")
plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize= (15,10))
sns.barplot(data = kategori_raporu,x = "kategori",y= "ortalama_musteri_puani")
plt.title("Kategoriye göre ortalama müşteri puanı")
plt.xlabel("Kategori")
plt.ylabel("Ortalama Musteri Puanı")
plt.tight_layout()
plt.show()
```

```python
## Hangi ödeme türü daha fazla getiri ve müsteri memmuniyeti sağlıyor?
odeme_musteri = df.groupby("odeme_turu").agg(
    ortalama_gelir =  ("toplam_tutar","mean"),
    ortalama_musteri_memmuniyeti = ("musteri_puani","mean")
)
odeme_musteri
```

```python
odeme_musteri = odeme_musteri.sort_values(by = "ortalama_gelir",ascending = False)
odeme_musteri
```

```python
## Odeme turu Grafiği
plt.Figure(figsize = (15,10))
sns.barplot(data = odeme_musteri,x = "odeme_turu",y ="ortalama_gelir")
plt.title("Odeme türüne göre toplam gelir")
plt.xlabel("Odeme Türü")
plt.ylabel("Gelir")
plt.tight_layout()
plt.show()
```

```python
## Odeme turu Grafiği
plt.Figure(figsize = (15,10))
sns.barplot(data = odeme_musteri,x = "odeme_turu",y ="ortalama_musteri_memmuniyeti")
plt.title("Odeme türüne göre toplam gelir")
plt.xlabel("Odeme Türü")
plt.ylabel("Gelir")
plt.tight_layout()
plt.show()
```
